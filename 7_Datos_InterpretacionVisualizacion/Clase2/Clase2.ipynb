Ejercicio Práctico: "Del Análisis al Relato Visual"

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns

np.random.seed(10)

df = pd.DataFrame({
    'fecha': pd.date_range('2025-01-01', periods=12, freq='M'),
    'region': np.random.choice(['Norte','Centro','Sur'], 12),
    'ventas': np.random.randint(2000,5000,12),
    'ticket_promedio': np.random.uniform(15,40,12),
    'reclamos': np.random.randint(20,150,12)
})

plt.figure(figsize=(8,4))

sns.lineplot(
    data=df,
    x='fecha',
    y='ventas',
    hue='region',
    marker='o'
)

plt.title('Evolución mensual de ventas por región – 2025')
plt.xlabel('Mes')
plt.ylabel('Ventas')
plt.xticks(rotation=45)
plt.show()
```

Tendencias de ventas

El análisis temporal de ventas durante 2025 muestra diferencias claras entre regiones. Algunas regiones presentan una tendencia creciente a lo largo del año, indicando una mejora sostenida en el desempeño comercial. Otras regiones mantienen niveles más estables o incluso fluctuaciones, lo que sugiere menor dinamismo en su crecimiento.


```python
plt.figure(figsize=(6,4))

sns.barplot(
    data=df,
    x='region',
    y='ticket_promedio',
    estimator=np.mean
)

plt.title('Ticket promedio por región – 2025')
plt.xlabel('Región')
plt.ylabel('Ticket promedio')
plt.show()
```

Explicación del desempeño

El análisis del ticket promedio revela que parte del crecimiento puede explicarse por el mayor gasto por transacción en ciertas regiones. Aquellas regiones con tickets promedio más altos logran mayores ingresos incluso con volúmenes similares de clientes. Esto sugiere oportunidades de replicar estrategias comerciales o de pricing en otras zonas.


```python
plt.figure(figsize=(6,4))

sns.scatterplot(
    data=df,
    x='reclamos',
    y='ventas',
    hue='region',
    size='ticket_promedio',
    sizes=(50,200)
)

plt.title('Relación entre reclamos y ventas')
plt.xlabel('Número de reclamos')
plt.ylabel('Ventas')
plt.show()
```

Calidad del servicio

Finalmente, al analizar la relación entre ventas y reclamos, se observa que algunas regiones logran altos niveles de ventas manteniendo niveles relativamente bajos de reclamos, lo que indica una experiencia de cliente más consistente. Sin embargo, en regiones donde el crecimiento de ventas viene acompañado de un aumento en reclamos, se recomienda revisar procesos operativos o de atención para evitar impactos en la satisfacción del cliente.

# Problema 1
Se hizo una encuesta a 325 personas mayores de 16 años y se encontró que 120 iban al teatro regularmente. Hallar, con un nivel de confianza del 94%, un intervalo para estudiar la proporción de los ciudadanos que van al teatro regularmente.

```python
# Datos
n = 325
x = 120
confianza = 0.94
```

```python
proporcionMuestral = x/n
alfa = 1 - confianza
z = stats.norm.ppf(1 - alfa/2)

print("Proporción muestral:", proporcionMuestral)
print("Alfa:", alfa)
print("Valor crítico z:", z)
```

```python
errorEstandar = np.sqrt(proporcionMuestral * (1 - proporcionMuestral) / n)

print("Error estándar:", errorEstandar)
```

```python
limiteInferior = proporcionMuestral - z * errorEstandar
limiteSuperior = proporcionMuestral + z * errorEstandar

print("Intervalo de confianza al 94%: [{:.4f}, {:.4f}]".format(limiteInferior, limiteSuperior))
```

Con un 94% de confianza, la proporción de ciudadanos que van al teatro regularmente está entre 32.2% y 41.6%

# Problema 2

```python
# Datos
n = 800
x = 240
confianza = 0.9545
```

```python
proporcionMuestral = x/n
alfa = 1 - confianza
z = stats.norm.ppf(1 - alfa/2)

print("Proporción muestral:", proporcionMuestral)
print("Alfa:", alfa)
print("Valor crítico z:", z)
```

```python
errorEstandar = np.sqrt(proporcionMuestral * (1 - proporcionMuestral) / n)

print("Error estándar:", errorEstandar)
```

```python
# Cálculo del intervalo de confianza
limiteInferior = proporcionMuestral - z * errorEstandar
limiteSuperior = proporcionMuestral + z * errorEstandar

print("Intervalo de confianza al 95.45%: [{:.4f}, {:.4f}]".format(limiteInferior, limiteSuperior))
```

Con un 95,45% de confianza, la proporción de intención de voto por el partido A está entre 26.8% y 33,2%

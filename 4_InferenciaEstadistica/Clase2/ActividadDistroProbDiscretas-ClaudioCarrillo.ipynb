# Problema 1
Un estudio determinó que al seleccionar al azar adultos que poseen teléfonos inteligentes, el 54% de ellos los usa estando en clase o en reuniones. Se quiere encontrar la probabilidad de que, seleccionando al azar 8 personas con teléfono inteligente, exactamente 6 de ellas los utilicen en clase o reuniones.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Evento: 6 personas usando teléfono inteligente

posibilidades = [1, 2, 3, 4, 5, 6, 7, 8]
n = 6

valores = np.random.choice(posibilidades, size=n)
valores, conteos = np.unique(valores, return_counts=True)
frecuencias_relativas = conteos / n
```

```python
for valor, frec_rel in zip(valores, frecuencias_relativas):
    print(f"Valor: {valor}, Frecuencia Relativa: {frec_rel*100,2}%")
```

```python
# Grafico de la distribución de frecuencias relativas
plt.bar(valores, frecuencias_relativas, color='skyblue')
plt.xlabel('Valor')
plt.ylabel('Frecuencia Relativa')
plt.title('Distribución de Frecuencias Relativas')
plt.xticks(valores)
```

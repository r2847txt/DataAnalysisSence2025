# Problema 1:
Usted va a jugar dados al casino, lanza dos dados y para ganar lo apostado, necesita que la suma esté en el intervalo [7, 9]. ¿Qué probabilidad tiene de ganar la apuesta?

```python
import numpy as np
import matplotlib.pyplot as plt
```

__Espacio muestral__: 6 * 6 = 36
__Combinaciones__:
- __Suma = 7__: (1,6), (2,5), (3,4), (4,3), (5,2), (6,1) → 6 combinaciones
- __Suma = 8__: (2,6), (3,5), (4,4), (5,3), (6,2) → 5 combinaciones
- __Suma = 9__: (3,6), (4,5), (5,4), (6,3) → 4 combinaciones

Total casos favorables: 6 + 5+ 4 = 15

```python
dados = range(1,7)
combinaciones = [(d1, d2) for d1 in dados for d2 in dados]
sumas = [sum(comb) for comb in combinaciones]
favorables = [s for s in sumas if 7 <= s <= 9]
probabilidad = len(favorables) / len(sumas)

print(f"Probabilidad teórica: {probabilidad}")
```

```python
n = 100000
dado1 = np.random.randint(1, 7, n)
dado2 = np.random.randint(1, 7, n)
sumas_simulacion = dado1 + dado2

ganados = np.sum((sumas_simulacion >= 7) & (sumas_simulacion <= 99))
prob_simulada = ganados / n

print(f"Probabilidad simulada (1000 lanzamientos): {prob_simulada}")
```

# Problema 2
Una empresa recibe llamadas telefónicas de sus clientes y el tiempo de espera entre cada llamada sigue una distribución exponencial, con una tasa de llegada promedio de 4 llamadas por hora. El jefe quiere saber cuál es la probabilidad de que la empresa tenga que esperar más de 10 minutos entre 2 llamadas consecutivas.

```python
from scipy.stats import expon
```

```python
llamadas_hora = 4
lam = llamadas_hora / 60
tiempo_limite = 10

prob_mas_10 = expon.sf(tiempo_limite, scale = 1/lam)
print(f"La probabilidad de esperr más de 10 min es: {prob_mas_10}")
```
